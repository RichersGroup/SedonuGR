```python
import h5py 
import numpy as np
```

```python
f = h5py.File("../../BH2_50ms/fluid_combined.h5")
x = np.array(f["axes/x0(cm)[mid]"])
y = np.array(f["axes/x1(cm)[mid]"])
z = np.array(f["axes/x2(cm)[mid]"])

T = np.array(f["T_gas(K,tet)"])
mu_nue = np.array(f["munue(erg)"])
```

```python
# core setup for radius 1 = 47.26 km 
# ix = 61
# iy = 61
# iz = 8

ix = 61
iy = 61
iz = 3

print(x[ix], y[iy], z[iz])
R = np.sqrt(x[ix]**2 + y[iy]**2 + z[iz]**2)
print(R/1e5, "in km")


core_T = T[ix,iy,iz]
core_mu_nue = mu_nue[ix,iy,iz]

print(core_mu_nue * 624150.934, "in MeV")
print(core_T, " in K; ", core_T/1e9, " in e9 K.")

```
